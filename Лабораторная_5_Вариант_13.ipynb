```python
from pathlib import Path
from getpass import getpass
from huggingface_hub import HfApi

folder = Path("/content/lab5_variant13")

files_to_upload = [
    "sms-spam.csv",
    "best_model.joblib",
    "metrics.csv",
    "metadata.json",
    "benchmark_payloads.json",
    "requirements-model.txt"
]

# Проверяем, что результаты обучения сохранились
missing = [
    name for name in files_to_upload
    if not (folder / name).is_file()
]

if missing:
    raise FileNotFoundError(
        "Не найдены файлы: "
        + ", ".join(missing)
        + ". Сначала выполни основную ячейку обучения."
    )

# Токен вводится скрыто, в код его вставлять не нужно
token = getpass("Вставь токен Hugging Face и нажми Enter: ")

try:
    api = HfApi(token=token)

    api.upload_folder(
        repo_id="sdsffdsasd/sms-spam-variant13",
        repo_type="model",
        folder_path=str(folder),
        allow_patterns=files_to_upload,
        commit_message="Upload SMS spam model and training results"
    )

    print("\nФайлы успешно загружены!")
    print("Репозиторий:")
    print("https://huggingface.co/sdsffdsasd/sms-spam-variant13")
    print("\nЗагруженные файлы:")

    for name in files_to_upload:
        print("✓", name)

finally:
    token = None
    api = None
```
